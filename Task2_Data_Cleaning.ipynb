In this task we clean and prepare the student dataset for analysis by handling missing values, duplicates, and formatting issues.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
print("Libraries imported!")
```

## 2. Load Dataset

```python
df = pd.read_csv('students.csv')
print(f"Original shape: {df.shape}")
df.head()
```

## 3. Check Missing Values (Before Cleaning)

```python
print("Missing values BEFORE cleaning:")
print(df.isnull().sum())
print()
missing_pct = (df.isnull().sum() / len(df)) * 100
print("Missing percentage:")
print(missing_pct[missing_pct > 0])
```

## 4. Handle Missing Values
The `MathScore` column has 2 missing values. We fill them with the column mean — a standard imputation technique.

```python
# Fill missing MathScore with mean
mean_math = df['MathScore'].mean()
print(f"Mean MathScore (used for filling): {mean_math:.2f}")

df['MathScore'].fillna(round(mean_math, 2), inplace=True)

print("\nMissing values AFTER filling:")
print(df.isnull().sum())
```

## 5. Check for Duplicate Rows

```python
duplicates = df.duplicated().sum()
print(f"Total duplicate rows: {duplicates}")

if duplicates > 0:
    print("Duplicate rows:")
    print(df[df.duplicated()])
    df.drop_duplicates(inplace=True)
    print(f"After removing duplicates: {df.shape}")
else:
    print("No duplicates found. Dataset is clean!")
```

## 6. Check & Fix Data Types

```python
print("Data types BEFORE:")
print(df.dtypes)

# StudentID should be treated as object (ID, not numeric)
df['StudentID'] = df['StudentID'].astype(str)

# Ensure scores are float
for col in ['MathScore', 'ScienceScore', 'EnglishScore', 'Attendance']:
    df[col] = df[col].astype(float)

print("\nData types AFTER:")
print(df.dtypes)
```

## 7. Fix & Validate Categorical Columns

```python
# Standardize Gender column (strip whitespace, capitalize)
df['Gender'] = df['Gender'].str.strip().str.capitalize()
print("Gender values:", df['Gender'].unique())

# Standardize Grade column
df['Grade'] = df['Grade'].str.strip().str.upper()
print("Grade values:", df['Grade'].unique())

# Standardize City column
df['City'] = df['City'].str.strip().str.title()
print("City values:", df['City'].unique())
```

## 8. Validate Score Ranges
Scores must be between 0 and 100, and attendance between 0 and 100.

```python
score_cols = ['MathScore', 'ScienceScore', 'EnglishScore', 'Attendance']

for col in score_cols:
    invalid = df[(df[col] < 0) | (df[col] > 100)]
    if len(invalid) > 0:
        print(f"WARNING: {col} has {len(invalid)} out-of-range values!")
    else:
        print(f"{col}: All values in valid range (0–100) ✓")
```

## 9. Validate Age Range

```python
print("Age range:", df['Age'].min(), "to", df['Age'].max())
invalid_age = df[(df['Age'] < 10) | (df['Age'] > 25)]
print(f"Invalid age entries: {len(invalid_age)}")
```

## 10. Add Computed Column: Average Score

```python
df['AvgScore'] = ((df['MathScore'] + df['ScienceScore'] + df['EnglishScore']) / 3).round(2)
print("AvgScore column added!")
print(df[['Name', 'MathScore', 'ScienceScore', 'EnglishScore', 'AvgScore']].head(10))
```

## 11. Final Clean Dataset Overview

```python
print(f"Final dataset shape: {df.shape}")
print(f"Total missing values: {df.isnull().sum().sum()}")
print()
df.describe()
```

## 12. Save Cleaned Dataset

```python
df.to_csv('students_cleaned.csv', index=False)
print("Cleaned dataset saved as 'students_cleaned.csv'")
```
