## 7. Identify Numeric vs Categorical Columns

```python
numeric_cols = df.select_dtypes(include='number').columns.tolist()
categorical_cols = df.select_dtypes(include='object').columns.tolist()

print("Numeric Columns:", numeric_cols)
print()
print("Categorical Columns:", categorical_cols)
```

## 8. Check for Missing Values

```python
print("Missing Values per Column:")
print(df.isnull().sum())
print()
print(f"Total Missing Values: {df.isnull().sum().sum()}")
```

## 9. Unique Values in Categorical Columns

```python
for col in categorical_cols:
    print(f"{col}: {df[col].nunique()} unique values → {df[col].unique()}")
    print()
```

## 10. Dataset Description

This dataset contains **academic performance records** of 30 students from various cities in Pakistan. Each record includes:

| Column | Description |
|--------|-------------|
| StudentID | Unique student identifier |
| Name | Student full name |
| Age | Age in years (15–18) |
| Gender | Male / Female |
| MathScore | Score out of 100 |
| ScienceScore | Score out of 100 |
| EnglishScore | Score out of 100 |
| Attendance | Attendance percentage |
| Grade | Final grade (A/B/C/D/F) |
| City | City of the student |

**Use case:** This data can help educators identify low-performing students, track attendance impact on grades, and compare subject-wise performance across cities.
